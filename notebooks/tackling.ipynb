```python
import pandas as pd
import numpy as np
import os
```

```python
parent_folder = "/run/media/saliherdemk/HIKVISION/osu-dataset-new/"
dfs = []
for folder in os.listdir(parent_folder):
    full_path = os.path.join(parent_folder, folder, "beatmaps.csv")
    dfs.append(pd.read_csv(full_path))
result = pd.concat(dfs, ignore_index=True)
```

```python
import matplotlib.pyplot as plt

def visualize(df):
    df["diff_rating_rounded"] = df["difficulty_rating"].astype("int")
    plt.hist(df["diff_rating_rounded"], bins=range(int(df["diff_rating_rounded"].min()), int(df["diff_rating_rounded"].max())+2), 
    edgecolor='black', align='left')
    plt.xlabel("Rounded Difficulty Rating")
    plt.ylabel("Count")
    plt.title("Distribution of Rounded Difficulty Ratings")
    plt.show()
```

```python
visualize(result)
```

```python
eval_df = pd.read_csv("/run/media/saliherdemk/HIKVISION/osu-dataset-new/eval-34/beatmaps.csv")
```

```python
visualize(eval_df)
```
