```python
import os
import shutil
from tqdm import tqdm

def split_into_subfolders(folder_path, batch_size=500):
    all_files = os.listdir(folder_path)
    chunks = [all_files[i: i + batch_size] for i in range(0, len(all_files), batch_size)]
    for idx, chunk in tqdm(enumerate(chunks), total=len(chunks)):
        desc_folder = f"/home/saliherdemk/osu-chunks/{idx}"
        os.makedirs(desc_folder, exist_ok=True)
        for f in chunk:
            source = os.path.join(songs_folder,f)
            destination = os.path.join(desc_folder,f)
            shutil.move(source, destination)

```
